```python
#CREATE DICTIONARY

new_dict = {
     1 : "Amit",
     2 : "Neha",
     3 : "Varad",
     4 : "Rahul",
     5 : "Shlok"

}

#ADDING DICTIONARY

print("Before adding", new_dict)
new_dict[6] = "Aditya" 
print("After adding", new_dict)

#DELETING DICTIONARY
del new_dict[1]



#UPDATE DICTIONARY
##Updating existing key
print("Before updating", new_dict)
new_dict.update({2 : "Anish"})
print("After updating", new_dict)

```

```python
new_list = ["Amit", "Neha", "Varad", "Rahul", "Shlok"]
print(new_list)
new_list.append("Anish")
print(new_list)
new_list.insert(1, "Aditya")
print(new_list)
new_list.pop()
print(new_list)
new_list.remove("Neha")
print(new_list)
```
