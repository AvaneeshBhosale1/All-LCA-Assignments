```python
#To verify is the sides are of a right-angles triangle

def right(a:float, b:float, c:float):
    
    if (a*a + b*b == c*c) or (a*a + c*c == b*b) or (c*c + b*b == a*a):
      print("Right Angled Triangle")
      return True
    else:
      print("Not Right Angled Triangle")
      return False

right(3,4,5)
right(3,4,9)

#OR

def right2(x,y,z):
    sides = [x.y.z]
    sides.sort()
    z = sides[2]#giving index to other arguments is not necessary.
    if(x*x + y*y == z*z):
      print("Right Angled Triangle")
      return True
    else:
      print("Not Right Angled Triangle")
      return False 

right(3,5,4)
right(3,9,4)
```
